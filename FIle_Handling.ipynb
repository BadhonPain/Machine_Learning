```python
import numpy as np
```

```python
file_path = "/content/drive/MyDrive/File_Test/marks.txt"
with open(file_path,"r") as file:
  contents = file.readlines()
  only_marks = []
  collections_of_lines = []
  for lines in contents:
    line = lines.split()
    # collections_of_lines.append(line)
    only_marks.append(int(line[1]))

  total_sum = 0
  plus = 0
  max_num = float('-inf')
  for nums in only_marks:
    if nums >= 80:
      plus+=1
    total_sum+=nums
    max_num = max(max_num,nums)

  print(f"Average Mark: {total_sum/len(only_marks)}\nHighest Mark: {max_num}\nStudents > 80: {plus}\n")
  # print(contents)
  # print(collections_of_lines)
  # print(only_marks)
  np_marks = np.array(only_marks)
  print(f"With Numpy:\nAverage: {np.mean(np_marks)}")
  print(f"Max: {np.max(np_marks)}")
  # print(f"Plus Count: {len([mark for mark in np_marks if mark >= 80])}")
  print(f"Plus Count: {(np_marks >= 80).sum()}")

# Commented lines are written for debugging
```
