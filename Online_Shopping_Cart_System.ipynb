```python
class ShoppingCart :
  def __init__(self,customer_name = "",items=None):
    self.__customer_name = customer_name
    self.__items = items if items is not None else []

  def add_item(self,price):
    if price >= 0:
      self.__items.append(price)
    else :
      print("Error: Invalid Price !")

  def remove_item(self,price):
    try :
      self.__items.remove(price)
    except ValueError:
      print(f"Error: {price} not in the items !")

  def calculate_total(self):
    total_sum = 0
    for prices in self.__items:
      total_sum +=prices
    return total_sum

  def apply_discount(self):
    payable = self.calculate_total()
    if payable >= 3000:
      payable -= payable*0.1
    return payable

  def display_cart(self):
    print(f"\nCustomer Name: {self.__customer_name}")
    print(f"Items in the Cart: {self.__items}")
    print(f"Total Items in the Cart: {len(self.__items)}")
    print(f"Total Amount: {self.calculate_total()}")
    print(f"Final Payable Amount: {self.apply_discount()}\n")

  def clear_cart(self):
    self.__items.clear()
    print("All the items of your cart removed successfully !")



cart1 = ShoppingCart("Messi")

cart1.add_item(1200)
cart1.add_item(1500)
cart1.add_item(800)
cart1.add_item(-1000)   # Trying to add Invalid item

cart1.display_cart()

cart1.remove_item(1500)
cart1.remove_item(10)   # Finding absent item
cart1.display_cart()

cart1.clear_cart()
cart1.display_cart()




```
