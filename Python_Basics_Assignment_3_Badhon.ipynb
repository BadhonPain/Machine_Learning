```python
# Given the list:
# fruits = ["apple", "banana", "cherry", "mango"]
# Use enumerate() to print each fruit with its index number.

# Output format example:
# 0 apple
# 1 banana
# 2 cherry
# 3 mango

fruits = ["apple", "banana", "cherry", "mango"]

for key, items in enumerate(fruits) :
  print(key,items)
```

```python
# Given two lists:
# students = ["Rahim", "Karim", "Ayesha"]
# scores = [85, 90, 78]

# Use zip() to print each student name along with their score.

# Example Output:
# Rahim scored 85
# Karim scored 90
# Ayesha scored 78

students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]

for items1, items2 in zip(students, scores):
  print(items1," scored ",items2)
```

```python
# Create a tuple of 5 subjects.
# Using a for loop, print each subject one by one.

t = ("CSE", "EEE", "ME", "BME", "ARCH")

for items in t:
  print(items)
```

```python
# Given the sets:
# set1 = {1, 2, 3, 4, 5}
# set2 = {4, 5, 6, 7}

# Print the common elements using set operation
# Print the unique elements of set1

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

print("Common Elements:",set1 & set2)
print("Unique Elemets in set1:", set1-set2)
```

```python
# Given the dictionary:
# marks = {
# "Math": 80,
# "Physics": 75,
# "Chemistry": 85
# }

# Using a loop and enumerate():
# ●Print the subject number, subject name, and marks
# Example Output:
# 1 Math: 80
# 2 Physics: 75
# 3 Chemistry: 85

marks = {
"Math": 80,
"Physics": 75,
"Chemistry": 85
}

for i,(key, value) in enumerate(marks.items(),1):
    print(i,key,':',value)

```
